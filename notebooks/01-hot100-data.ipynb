This is actually a famous historical anomaly (1990) in the Billboard charts:

**Unchained Melody by The Righteous Brothers appears twice on the Hot 100.**

The *1965 original* used in the film Ghost, and a *newly recorded version* released by a different label following the movie's success were considered distinct recordings.

But for the purpose of lyrics finding, they shall be treated as the same song. And the song's popularity features need to be integrated.

```python
hot100_song_df = hot100_song_df.sort_values(["title", "performer", "chart_week", "wks_on_chart"])
hot100_song_df
```

```python
hot100_song_df = (
    hot100_song_df
    .groupby(["title", "performer", "chart_week"], as_index=False)
    .agg(
        wks_on_chart=("wks_on_chart", "max"),
        peak_pos=("peak_pos", "min"),
    )
)
```

```python
hot100_song_df.shape
```

The duplicates are removed.

```python
hot100_song_df[(hot100_song_df["title"] == "Unchained Melody") & (hot100_song_df["performer"] == "The Righteous Brothers")]
```

But problems presisted:
* from 1965-10-09 to 1990-08-25, "wks_on_chart" jumps but "peak_pos" does not.
* from 1990-12-29 to 1991-01-05, "wks_on_chart" jumps from 19 to 14
* from 1990-12-29 to 1991-01-05, "peak_pos" jumps from 4 to 19

In this case, it is better to understand both "wks_on_chart" and "peak_pos" to preserve memory, so each of them can be resolved with cumcount() and cummin(). 

This method can actually make the meanings of "wks_on_chart" and "peak_pos" consistent (with past memory) across the entire DataFrame.

```python
# Make sure rows are chronological within each song
hot100_song_df = hot100_song_df.sort_values(["title", "performer", "chart_week"])

# Rebuild weeks on chart: 1, 2, 3, ...
hot100_song_df["wks_on_chart"] = hot100_song_df.groupby(["title", "performer"]).cumcount() + 1

# Rebuild peak position: best position achieved up through that week
hot100_song_df["peak_pos"] = hot100_song_df.groupby(["title", "performer"])["peak_pos"].cummin()
```

Great!

For the purpose of lyrics finding, "title" and "performer" are enough to define a song. All other attributes ("chart_week", "wks_on_chart", "peak_pos") just indicate a song's popularity. 

Here, I would like to integrate those attributes and generate popularity features for each song:

* **chart_weeks**: a list of all weeks on chart
* **periods_on_chart (not for MVP now)**: number of distinct periods in which a song appears on the Hot 100. A re-entry after only a few weeks may simply reflect a brief interruption, while a return after a longer absence can indicate renewed popularity. Need info from hot100_chart_df.
* **wks_on_chart**: the total number of weeks on chart
* **peak_pos**: the best ever position on chart

```python
hot100_song_df = (
    hot100_song_df
    .sort_values(["title", "performer", "chart_week"])
    .groupby(["title", "performer"], as_index=False)
    .agg(
        chart_weeks=("chart_week", list), # ordered from earlier sort_values
        wks_on_chart=("wks_on_chart", "max"),
        peak_pos=("peak_pos", "min"),
    )
)
```
